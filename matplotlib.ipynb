## Plot customization

### marker
* `'o'`: **Circle**
* `'s'`: **Square**
* `'^'`: **Triangle**
* `'x'`: **X mark**
* `'*'`: **Star**

### markersize or ms
* `4`: **Small** (subtle, clean for dense data)
* `6`: **Default** (standard size for most plots)
* `8`: **Medium** (clear and highly visible)
* `10`: **Large** (stands out, good for highlights)
* `12`: **Extra Large** (used for emphasis or key data points)

### markerfacecolor or mfc
* `'b'`, `'r'`, `'g'`: **Standard Color**
* `'none'`: **Hollow Fill**
* `'white'`: **White Fill**
* `'auto'`: **Line-Matching Color**

### markeredgecolor or mec
* `'k'`: **Black Outline** (Creates a crisp border that makes points pop)
* `'w'`: **White Outline** (Creates visual space in crowded, overlapping plots)
* `'none'`: **No Outline** (Removes the border completely for a smoother look)
* `'b'`, `'r'`, `'g'`: **Standard Color** (Applies a specific color border to the marker)


### color
* `'b'`: **Blue**
* `'r'`: **Red**
* `'g'`: **Green**
* `'c'`: **Cyan**
* `'k'`: **Black**

### linewidth or lw
* `1`: **Thin** (subtle, good for dense or multiple lines)
* `1.5`: **Default** (standard baseline thickness)
* `2`: **Medium** (clear and highly readable)
* `3`: **Thick** (emphasizes a primary data line)
* `4`: **Extra Thick** (used for major trends or highlights)


### linestyle or ls
* `'-'`: **Solid** (the default unbroken line)
* `'--'`: **Dashed** (ideal for secondary data or averages)
* `':'`: **Dotted** (commonly used for reference grids or thresholds)
* `'-.'`: **Dash-dot** (alternating dashes and dots)

### alpha
* `1.0`: **Solid** (fully opaque; the default setting)
* `0.7`: **Slightly Translucent** (reduces glare on overlapping lines)
* `0.5`: **Semi-Transparent** (perfect for background data trends)
* `0.3`: **Lightly Transparent** (great for dense scatter points or confidence intervals)
* `0.1`: **Very Faint** (used for minor grid lines or massive datasets)


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([20, 30, 35, 15])
y2 = np.array([39, 22, 41, 6])
y3 = np.array([30, 53, 31, 4])

line_style = dict(    
    marker="d",
    markersize=6, 
    markerfacecolor="red", 
    markeredgecolor="black",
    linestyle="-." ,
    linewidth="3"
    )
```

```python
plt.plot(x, y1, color="blue", alpha=0.5, **line_style)

plt.plot(x, y2, color="black", alpha=0.2, **line_style)

plt.plot(x, y3, color="yellow", **line_style)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([20, 30, 35, 15])
y2 = np.array([39, 22, 41, 6])
y3 = np.array([30, 53, 31, 4])
```

```python
plt.title(
    "Class Size",
    fontsize=25,
    family="Arial",
    fontweight="bold",
    color="Red")

plt.xlabel(
    "Years",
    fontsize=20,
    family="Arial",
    fontweight="bold",
    color="Green")

plt.ylabel(
    "Students",
    fontsize=20,
    family="Arial",
    fontweight="bold",
    color="orange")

plt.tick_params(
    axis="x",
    colors="blue"
)


plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)

plt.xticks(x)
plt.show()
```

## Grid Lines, Legend, and Limiting

### plt.grid()
* It helps to make easier to read by adding referece lines
* axis="both"
* linewidth=2
* color="lightgray"
* linestyle="--"


### plt.legend()

### plg.xlim()

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.array([2023, 2024, 2025, 2026])
y = np.array([20, 30, 35, 200])

plt.grid(
    axis="y",
    linewidth=2,
    color="lightgray",
    linestyle=":"
)

plt.plot(x, y, label="students")
plt.legend(loc="upper left")
plt.xlim(2023, 2024)
plt.ylim(0, 50)
plt.xticks(x)
plt.show()
```

## Pie Chart
- Circular chart divided into slices to show percentages of the total.
- Good for visualizing distribution among categories
- Univariate/Bivariate Analysis
- Categorical VS numerical
- Use Case - To find  contibution on a standard scale

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
categories = np.array(["Freshmen", "Sophomores", "Juniors", "Seniors"])
values = np.array([300, 250, 258, 242])
colors = ["red", "yellow", "green", "blue"]
```

```python
plt.pie(
    values,
    labels=categories,
    autopct="%1.1f%%",
    colors=colors,
    explode=[0, 0, 0, 0.1],
    shadow=True,
    startangle=90
    )

plt.title("College")
plt.show()
```

### Scatter Plots
- shows the relationship between two variables
- helps to identify a correlation (+, -, None)
- Example: study hours vs. Test scores
- Bivariate Analysis
- Nummerical vs Numerical
- Use case - Finding correlation


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x1 = np.array([0, 1, 1, 2, 3, 5, 4, 6, 7, 8, 8, 9]) # Hours studied
y1 = np.array([55, 60, 62, 70, 76, 80, 84, 86, 87, 90, 89, 93]) # Grandes

x2 = np.array([0, 2, 2, 4, 5, 5, 9, 3, 7, 4, 2, 9]) # Hours studied
y2 = np.array([55, 63, 66, 71, 76, 69, 91, 59, 87, 60, 24, 94]) # Grandes
```

## Histogram
- A visual representation of the distribution of quantitaive data.
- They group values into bins (intervals)
- and counts how many fall in each range.
- Univariate Analysis
- Numerical Col
- Use Case - Frequency Count

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
scores = np.random.normal(loc=80, scale=10, size=100)
scores = np.clip(scores, 0, 100)
```

```python
plt.hist(
    scores, 
    bins=10, 
    color="lightblue",
    edgecolor="black",
    )

plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("No. of studenst")

plt.show()
```

## Colored Scatter Plot

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("iris")
df.sample(5)
```

```python
df["species"] = df["species"].replace({"setosa":0, "versicolor":1, "virginica":2})
df.sample(5)
```

### cmap

* **`viridis`**: Continuous data, heatmaps, and general continuous plotting (default).
* **`coolwarm`**: Diverging data with a neutral center (e.g., correlations, temperature anomalies).
* **`tab10`**: Distinct categorical data with no intrinsic order (e.g., bar charts, different groups) [Python matplotlib.pyplot.colormaps].
* **`magma`**: High-density data plots and dark-themed visualizations.
* **`RdBu`**: Diverging values highlighting extreme positive and negative deviations.
* **`plasma`**: Vibrant continuous plots, medical imaging, and physics graphs.
* **`inferno`**: High-contrast continuous tracking mimicking thermal or heat energy.
* **`Blues`**: Sequential single-hue tracking (e.g., density scales, geographical elevation).
* **`turbo`**: Engineering and radar visualization requiring ultra-high rainbow contrast.
* **`Greys`**: Background grids, spatial layers, or secondary data configurations.

### colorbar

* **`viridis`**: Continuous data, heatmaps, and general continuous plotting (default). Includes a smooth purple-to-yellow colorbar.
* **`coolwarm`**: Diverging data with a neutral center (e.g., correlations, anomalies). Colorbar shows a clear blue-to-red gradient with white in the middle.
* **`tab10`**: Distinct categorical data with no intrinsic order (e.g., bar charts, groups) [Python matplotlib.pyplot.colormaps]. Uses a discrete colorbar with 10 solid, separate color blocks.
* **`magma`**: High-density data plots and dark-themed visualizations. Colorbar ranges from deep black through purple to bright white-pink.
* **`RdBu`**: Diverging values highlighting extreme deviations. Colorbar scales from intense dark red to dark blue.
* **`plasma`**: Vibrant continuous plots and physics graphs. Colorbar shifts sharply from dark blue to neon pink and yellow.
* **`inferno`**: High-contrast continuous tracking mimicking heat energy. Colorbar tracks from pure black to bright golden yellow.
* **`Blues`**: Sequential single-hue tracking (e.g., density scales). Colorbar transitions monotonically from light faint blue to dark navy.
* **`turbo`**: Engineering and radar visualization. Colorbar provides a smooth, fully visible rainbow spectrum (blue, green, yellow, red).
* **`Greys`**: Background grids or secondary data configurations. Colorbar displays a standard, clean white-to-black grayscale.



```python
plt.scatter(
    df["sepal_length"], 
    df["petal_length"],
    c = df["species"],
    cmap="tab10",
    alpha=0.5
    
    )

plt.title("Iris", color="skyblue", fontsize=18, fontweight="bold")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.colorbar()
plt.show()
```

## Annotations

```python
import numpy as np
import matplotlib.pyplot as plt

# data points
x = np.array([2023, 2024, 2025, 2026])
y = np.array([20, 30, 35, 15])
z = np.array(["Launch", "Growth", "Peak", "Current"])

# Create the base scatter plot
plt.scatter(x, y, color="tab:red", s=100, zorder=3,)

# Loop through the arrays to annotate each point individually
for xi, yi, zi in zip(x, y, z):
    plt.text(
        x=xi, 
        y=yi + 1,          # Offsets the text slightly above the point
        s=zi,              # The text string from your z list
        ha='center',       # Centers the text horizontally over the point
        va='bottom',       # Aligns text boundary box to the bottom
        fontsize=10,
        fontweight='bold',
        color="black"
    )


plt.xlim(2022, 2027)
plt.ylim(10, 40)
plt.xticks(x)
plt.title("Yearly Progress with Annotations", fontsize=18)
plt.xlabel("Year")
plt.ylabel("Values")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.axhline(25, color="tab:blue", linestyle="-.")
plt.axvline(2024.5, color="tab:blue", linestyle="-.")

plt.show()
```

## Subplots

- Figure = The entire canvas
- Ax = A single plot (subplot)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# plt.subplots(rows, columns)

x = np.array([1, 2, 3, 4, 5])
figure , axes = plt.subplots(2, 2)
axes[0, 0].plot(x, x*2, color="red")
axes[0, 0].set_title("x*2")
axes[0, 0].set_xlabel("x")
axes[0, 0].set_ylabel("x*2")


axes[0, 1].bar(x, x**2, color="green")
axes[0, 1].set_title("x**2")

axes[1, 0].scatter(x, x**3, color="blue", marker="*")
axes[1, 0].set_title("x**3")

axes[1, 1].barh(x, x**4, color="purple")
axes[1, 1].set_title("x**4")


plt.tight_layout()
plt.show()
```

```python
data = np.array([[50, 10,  5],
                 [ 8, 45, 12],
                 [ 4,  7, 60]])

labels = ['Car', 'SUV', 'Truck']

plt.imshow(data, cmap="Blues")
plt.title("Vehicle Classifier Confusion Matrix")
plt.xlabel("Predicted Vehicle Type")
plt.ylabel("True Vehicle Type")
plt.xticks(range(3), labels)
plt.yticks(range(3), labels)

plt.colorbar(label="Count")
plt.show()
```

## Pandas + Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("movies.csv")
df.sample(5)
```

```python
df["release_date"] = pd.to_datetime(df["release_date"])
year_count = df["release_date"].dt.year.astype("Int16").value_counts(ascending=True)
year_count
```

```python
plt.bar(year_count.index, year_count.values, color="lightgreen", edgecolor="green")

plt.title("Movies")
plt.xlabel("Year")
plt.ylabel("# Movies")
plt.show()
```
